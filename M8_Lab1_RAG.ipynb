    <h2>🖨️ Pretty Print Function</h2>
</div>

<div class="description-box">
    <p style="margin: 0 0 8px 0;">The <span class="highlight">pretty_print()</span> function enhances output readability by transforming standard text into styled HTML blocks. This utility function replaces basic print statements with visually appealing formatted displays that improve the user experience when viewing model responses and system outputs.</p>
    
    <p style="margin: 8px 0 0 0;">Key features include automatic detection and formatting of bulleted lists, proper line break handling, and consistent visual styling that matches the laboratory's design theme. The function accepts two parameters: the text content to display and an optional title that appears as a header above the formatted output.</p>
</div>

<div class="code-note">
    <strong>Usage:</strong> Replace standard <code>print()</code> statements with <code>pretty_print()</code> throughout your notebook to maintain consistent, professional output formatting.
</div>

</body>
</html>

````python
# 🖨️ pretty_print(): Reusable HTML display function for model outputs
def pretty_print(text, title="🤖 Model Response"):
    """
    Display model response in styled HTML block.
    Handles bulleted lists and line breaks.
    """
    lines = text.strip().split('\n')
    is_bulleted = all(line.strip().startswith(("-", "•", "*")) for line in lines if line.strip())

    if is_bulleted:
        list_items = ''.join(f"<li>{line.lstrip('-•* ').strip()}</li>" for line in lines if line.strip())
        content_html = f"<ul style='margin-top: 6px;'>{list_items}</ul>"
    else:
        content_html = text.replace("\n", "<br>")  # fallback for plain lines

    display(HTML(f"""
    <div style="background-color:#f8f9fc; border-left:5px solid #4285f4;
                padding:16px; margin-top:16px; font-family:'Segoe UI', sans-serif;
                color:#202124; line-height:1.6;">
      <strong>{title}</strong><br><br>
      {content_html}
    </div>
    """))
````

    <h3>Understanding the RAG Architecture</h3>
    <p>The Retrieval-Augmented Generation system enhances language model responses by incorporating document-specific context. Unlike standard LLMs that rely solely on training data, RAG systems actively search through your documents to find relevant information before generating responses.</p>
</div>

<div class="explanation-box" style="background-color: #f8f9fa;">
    <h3>How Retrieval Works</h3>
    <p>When a user submits a query, the system:</p>
    <ol style="margin: 8px 0 0 20px; padding: 0;">
        <li>Converts the query into an embedding vector using the same model</li>
        <li>Searches the FAISS index for the k most similar document chunks</li>
        <li>Passes these relevant chunks as context to the language model</li>
        <li>Generates a response grounded in the retrieved information</li>
    </ol>
    <p style="margin-top: 12px;">This approach ensures responses are based on your specific documents rather than general knowledge, significantly improving accuracy and relevance.</p>
</div>

    <h3>📥 Task 1: Document Loading</h3>
    <div class="task-details">
        <p style="margin: 0 0 8px 0;">This task retrieves the course syllabus PDF from Dropbox and prepares it for RAG processing. The document undergoes three key transformations to enable efficient semantic search:</p>
        <ul class="process-list">
            <li><span class="highlight">Download:</span> Fetch the PDF file using the provided Dropbox URL</li>
            <li><span class="highlight">Load:</span> Extract text content from all pages using PyPDFLoader</li>
            <li><span class="highlight">Chunk:</span> Split the document into 1000-character segments with 200-character overlap to preserve context boundaries</li>
        </ul>
        <p style="margin: 8px 0 0 0;">The chunking strategy ensures that related information remains together while creating appropriately sized segments for embedding generation. The overlap prevents important context from being lost at chunk boundaries.</p>
    </div>

````python
# ++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
#>> DOCUMENT LOADING AND PROCESSING
# This cell downloads the PDF from Dropbox, loads it into memory,
# and splits it into manageable chunks for vector search
# ++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# 📦 Document Loaders
from langchain.document_loaders import PyPDFLoader  # Load content from PDFs
from langchain_community.document_loaders import TextLoader  # Load plain text files

# ✂️ Text Processing & Chunking
from langchain.text_splitter import RecursiveCharacterTextSplitter # Import the text splitter here


# Download PDF from Dropbox
dropbox_url = "https://www.dropbox.com/scl/fi/zedqrdppb6et1sm3s09r6/IE_5250_Applied_Generative_AI-2025.pdf?rlkey=tn3130kcd5o03twalmydn8t6p&e=1&dl=1"
pdf_path = "/content/document.pdf"

response = requests.get(dropbox_url)
with open(pdf_path, "wb") as file:
    file.write(response.content)

# Load and process the PDF
loader = PyPDFLoader(pdf_path)
documents = loader.load()

# Split into chunks for better retrieval accuracy
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    length_function=len,
    separators=["\n\n", "\n", " ", ""]
)
docs = text_splitter.split_documents(documents)

pretty_print(f"PDF successfully downloaded and processed\n{len(documents)} pages converted into {len(docs)} searchable chunks",
             title="📥 Document Loading Complete")
````

````python
# ++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
#>> EMBEDDING GENERATION AND VECTOR STORE CREATION
# This cell converts text chunks into vector embeddings using OpenAI's
# model and stores them in a FAISS index for fast similarity search
# ++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# ✂️ Text Processing & Chunking
from langchain.text_splitter import RecursiveCharacterTextSplitter  # Split text into chunks

# 📚 Vector Store & Embeddings
from langchain.vectorstores import FAISS  # FAISS for fast vector search

# Initialize embedding model
embedding_model = OpenAIEmbeddings(model="text-embedding-ada-002")

# Create FAISS vector store
vector_db = FAISS.from_documents(docs, embedding_model)

# Prepare sample chunks display
sample_chunks = []
for i in range(min(3, len(docs))):
    chunk_preview = docs[i].page_content[:150].strip()
    sample_chunks.append(f"• Chunk {i+1}: {chunk_preview}...")

sample_text = f"Embeddings successfully created for {len(docs)} chunks\n\nSample chunks:\n" + "\n".join(sample_chunks)
pretty_print(sample_text, title="🧠 Embedding Generation Complete")
````

````python
# ++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
#>> RETRIEVAL AND QUESTION ANSWERING
# This cell sets up the RAG chain, performs retrieval testing,
# and executes queries against the document
# ++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# 🔁 Retrieval-Augmented Generation (RAG)
from langchain.chains import RetrievalQA  # Combine retriever + LLM into a QA system

# Initialize language model
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)

# Create retriever
retriever = vector_db.as_retriever(search_kwargs={"k": 4})

# Test retrieval functionality
test_docs = retriever.get_relevant_documents("document")
retrieval_status = f"Retrieval system operational: {len(test_docs)} documents successfully retrieved"

# Build RAG chain
rag_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)

# Execute query
query = "What do I learn in the GenAI course in 3 bullets, software, appliciaon. Also who is the prof? any hitns for me to gain a good grade?"
result = rag_chain({"query": query})

# Format the complete response
query_result_text = f"{retrieval_status}\n\nQuery: {query}\n\nAnswer:\n{result['result']}\n\nSource Documents Used: {len(result['source_documents'])}"
pretty_print(query_result_text, title="🔍 RAG Query Results")
````

````python
# ============================================================================
# All Required Libraries
# ============================================================================

# System and utilities
import time
import os
import requests
import pandas as pd
from typing import List, Dict, Any
import numpy as np

# Core LangChain imports (existing)
from langchain.document_loaders import PyPDFLoader, CSVLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain.chains import RetrievalQA
from langchain.schema import Document

# FAISS (existing)
from langchain.vectorstores import FAISS

# NEW: ChromaDB and HuggingFace imports
import chromadb
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from sentence_transformers import SentenceTransformer

print("✅ All imports loaded successfully!")
print("🔸 FAISS + OpenAI: Ready")
print("🔸 ChromaDB + HuggingFace: Ready")
````

````python
# ============================================================================
# CONFIGURATION CLASS & UTILITY FUNCTIONS
# ============================================================================

class RAGConfig:
    """Configuration for RAG implementations"""

    # Document processing
    CHUNK_SIZE = 1000
    CHUNK_OVERLAP = 200

    # Retrieval settings
    TOP_K_DOCUMENTS = 4

    # Model configurations
    OPENAI_MODEL = "gpt-3.5-turbo"
    OPENAI_EMBEDDING_MODEL = "text-embedding-ada-002"
    HUGGINGFACE_EMBEDDING_MODEL = "all-MiniLM-L6-v2"  # Fast and effective

    # Test query (same as your original)
    TEST_QUERY = """What do I learn in the GenAI course in 3 bullets, software, application.
                    Also who is the prof? any hints for me to gain a good grade?"""

    # Data URLs
    PDF_URL = "https://www.dropbox.com/scl/fi/zedqrdppb6et1sm3s09r6/IE_5250_Applied_Generative_AI-2025.pdf?rlkey=tn3130kcd5o03twalmydn8t6p&e=1&dl=1"

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def download_file(url: str, filename: str) -> str:
    """Download file from URL"""
    response = requests.get(url)
    with open(filename, "wb") as file:
        file.write(response.content)
    return filename

def load_and_chunk_documents(file_path: str, file_type: str = "pdf") -> List[Document]:
    """Load and chunk documents"""
    # Load documents
    if file_type == "pdf":
        loader = PyPDFLoader(file_path)
    elif file_type == "csv":
        loader = CSVLoader(file_path, encoding="utf-8")
    else:
        raise ValueError(f"Unsupported file type: {file_type}")

    documents = loader.load()

    # Split into chunks
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=RAGConfig.CHUNK_SIZE,
        chunk_overlap=RAGConfig.CHUNK_OVERLAP,
        length_function=len,
        separators=["\n\n", "\n", " ", ""]
    )

    chunks = text_splitter.split_documents(documents)
    return chunks

def measure_time(func):
    """Decorator to measure execution time"""
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        return result, end_time - start_time
    return wrapper

pretty_print("Configuration and utilities loaded successfully!", title="⚙️ Setup Complete")
````

````python
# ============================================================================
# IMPLEMENTATION 1: FAISS + OPENAI (BASELINE)
# ============================================================================

class FAISSOpenAIRAG:
    """RAG implementation using FAISS and OpenAI embeddings"""

    def __init__(self):
        self.embedding_model = None
        self.vector_store = None
        self.rag_chain = None
        self.llm = ChatOpenAI(
            model=RAGConfig.OPENAI_MODEL,
            temperature=0
        )

    @measure_time
    def setup_embeddings(self):
        """Initialize OpenAI embeddings"""
        self.embedding_model = OpenAIEmbeddings(
            model=RAGConfig.OPENAI_EMBEDDING_MODEL
        )
        return "OpenAI embeddings initialized"

    @measure_time
    def create_vector_store(self, documents: List[Document]):
        """Create FAISS vector store"""
        self.vector_store = FAISS.from_documents(documents, self.embedding_model)
        return f"FAISS vector store created with {len(documents)} documents"

    @measure_time
    def setup_retrieval_chain(self):
        """Setup retrieval QA chain"""
        retriever = self.vector_store.as_retriever(
            search_kwargs={"k": RAGConfig.TOP_K_DOCUMENTS}
        )

        self.rag_chain = RetrievalQA.from_chain_type(
            llm=self.llm,
            chain_type="stuff",
            retriever=retriever,
            return_source_documents=True
        )
        return "FAISS retrieval chain setup complete"

    @measure_time
    def query(self, question: str):
        """Execute query and return results"""
        result = self.rag_chain({"query": question})
        return {
            'answer': result['result'],
            'source_docs': len(result['source_documents']),
            'sources': result['source_documents']
        }

pretty_print("FAISS + OpenAI implementation class defined successfully!", title="🔵 FAISS + OpenAI Ready")
````

````python
# ============================================================================
# IMPLEMENTATION 2: CHROMADB + HUGGINGFACE (ALTERNATIVE)
# ============================================================================

class ChromaHuggingFaceRAG:
    """RAG implementation using ChromaDB and HuggingFace embeddings"""

    def __init__(self):
        self.embedding_model = None
        self.vector_store = None
        self.rag_chain = None
        self.llm = ChatOpenAI(
            model=RAGConfig.OPENAI_MODEL,
            temperature=0
        )
        self.chroma_client = None

    @measure_time
    def setup_embeddings(self):
        """Initialize HuggingFace embeddings"""
        self.embedding_model = HuggingFaceEmbeddings(
            model_name=RAGConfig.HUGGINGFACE_EMBEDDING_MODEL,
            model_kwargs={'device': 'cpu'},  # Use 'cuda' if GPU available
            encode_kwargs={'normalize_embeddings': True}
        )
        return "HuggingFace embeddings initialized"

    @measure_time
    def create_vector_store(self, documents: List[Document]):
        """Create ChromaDB vector store"""
        # Initialize ChromaDB client
        self.chroma_client = chromadb.Client()

        # Create collection (delete if exists for fresh start)
        try:
            self.chroma_client.delete_collection("rag_collection")
        except:
            pass

        self.vector_store = Chroma.from_documents(
            documents=documents,
            embedding=self.embedding_model,
            collection_name="rag_collection",
            client=self.chroma_client
        )
        return f"ChromaDB vector store created with {len(documents)} documents"

    @measure_time
    def setup_retrieval_chain(self):
        """Setup retrieval QA chain"""
        retriever = self.vector_store.as_retriever(
            search_kwargs={"k": RAGConfig.TOP_K_DOCUMENTS}
        )

        self.rag_chain = RetrievalQA.from_chain_type(
            llm=self.llm,
            chain_type="stuff",
            retriever=retriever,
            return_source_documents=True
        )
        return "ChromaDB retrieval chain setup complete"

    @measure_time
    def query(self, question: str):
        """Execute query and return results"""
        result = self.rag_chain({"query": question})
        return {
            'answer': result['result'],
            'source_docs': len(result['source_documents']),
            'sources': result['source_documents']
        }

pretty_print("ChromaDB + HuggingFace implementation class defined successfully!", title="🟡 ChromaDB + HuggingFace Ready")
````

````python
# ============================================================================
# COMPARISON FRAMEWORK
# ============================================================================

class RAGComparator:
    """Compare different RAG implementations"""

    def __init__(self):
        self.results = {}

    def run_comparison(self, documents: List[Document], query: str):
        """Run comprehensive comparison between implementations"""

        pretty_print("Starting RAG Implementation Comparison...", title="🔄 Comparison in Progress")

        # Test FAISS + OpenAI
        print("\n🟢 Testing FAISS + OpenAI Implementation:")
        faiss_rag = FAISSOpenAIRAG()
        faiss_results = self._test_implementation(faiss_rag, documents, query, "FAISS+OpenAI")

        # Test ChromaDB + HuggingFace
        print("\n🟡 Testing ChromaDB + HuggingFace Implementation:")
        chroma_rag = ChromaHuggingFaceRAG()
        chroma_results = self._test_implementation(chroma_rag, documents, query, "ChromaDB+HuggingFace")

        # Store results
        self.results = {
            'FAISS+OpenAI': faiss_results,
            'ChromaDB+HuggingFace': chroma_results
        }

        # Generate comparison report
        self._generate_comparison_report()
        return self.results

    def _test_implementation(self, rag_impl, documents: List[Document], query: str, impl_name: str):
        """Test a single RAG implementation"""
        results = {'implementation': impl_name}

        try:
            # Setup embeddings
            _, embed_time = rag_impl.setup_embeddings()
            results['embedding_setup_time'] = embed_time
            print(f"   ✅ Embeddings setup: {embed_time:.2f}s")

            # Create vector store
            _, vector_time = rag_impl.create_vector_store(documents)
            results['vector_store_time'] = vector_time
            print(f"   ✅ Vector store creation: {vector_time:.2f}s")

            # Setup retrieval chain
            _, chain_time = rag_impl.setup_retrieval_chain()
            results['chain_setup_time'] = chain_time
            print(f"   ✅ Retrieval chain setup: {chain_time:.2f}s")

            # Execute query
            query_result, query_time = rag_impl.query(query)
            results['query_time'] = query_time
            results['answer'] = query_result['answer']
            results['source_docs_used'] = query_result['source_docs']
            print(f"   ✅ Query execution: {query_time:.2f}s")

            # Calculate total setup time
            results['total_setup_time'] = embed_time + vector_time + chain_time
            results['success'] = True

        except Exception as e:
            print(f"   ❌ Error in {impl_name}: {str(e)}")
            results['error'] = str(e)
            results['success'] = False

        return results

    def _generate_comparison_report(self):
        """Generate detailed comparison report"""
        print("\n" + "=" * 80)
        print("📊 COMPREHENSIVE RAG COMPARISON REPORT")
        print("=" * 80)

        # Setup Complexity Comparison
        print("\n🔧 SETUP COMPLEXITY:")
        print("-" * 40)

        for impl_name, results in self.results.items():
            if results['success']:
                print(f"\n{impl_name}:")
                print(f"  • Embedding Setup:     {results['embedding_setup_time']:.2f}s")
                print(f"  • Vector Store Setup:  {results['vector_store_time']:.2f}s")
                print(f"  • Chain Setup:         {results['chain_setup_time']:.2f}s")
                print(f"  • Total Setup Time:    {results['total_setup_time']:.2f}s")

        # Processing Speed Comparison
        print("\n⚡ PROCESSING SPEED:")
        print("-" * 40)

        successful_results = [impl for impl in self.results.values() if impl['success']]

        if len(successful_results) >= 2:
            fastest_setup = min(successful_results, key=lambda x: x['total_setup_time'])['implementation']
            fastest_query = min(successful_results, key=lambda x: x['query_time'])['implementation']

            for impl_name, results in self.results.items():
                if results['success']:
                    setup_indicator = "🏆" if impl_name == fastest_setup else "  "
                    query_indicator = "🏆" if impl_name == fastest_query else "  "

                    print(f"\n{impl_name}:")
                    print(f"  {setup_indicator} Setup Speed:  {results['total_setup_time']:.2f}s")
                    print(f"  {query_indicator} Query Speed:  {results['query_time']:.2f}s")
                    print(f"    Sources Retrieved: {results['source_docs_used']} documents")

pretty_print("Comparison framework ready!", title="📊 RAGComparator Loaded")
````

````python
# ============================================================================
# DETAILED ANALYSIS & RECOMMENDATIONS
# ============================================================================

def generate_detailed_recommendations(results):
    """Generate detailed recommendations based on comparison results"""

    print("\n" + "="*80)
    print("💡 DETAILED ANALYSIS & RECOMMENDATIONS")
    print("="*80)

    if not all(result['success'] for result in results.values()):
        print("⚠️ Some implementations failed. Check error messages above.")
        return

    faiss_results = results['FAISS+OpenAI']
    chroma_results = results['ChromaDB+HuggingFace']

    # Speed Analysis
    print("\n⚡ SPEED ANALYSIS:")
    print("-" * 50)

    setup_speedup = chroma_results['total_setup_time'] / faiss_results['total_setup_time']
    query_speedup = chroma_results['query_time'] / faiss_results['query_time']

    print(f"Setup Time Comparison:")
    print(f"  FAISS+OpenAI:        {faiss_results['total_setup_time']:.2f}s")
    print(f"  ChromaDB+HuggingFace: {chroma_results['total_setup_time']:.2f}s")

    if setup_speedup > 1:
        print(f"  → FAISS is {setup_speedup:.1f}x faster for setup")
    else:
        print(f"  → ChromaDB is {1/setup_speedup:.1f}x faster for setup")

    print(f"\nQuery Time Comparison:")
    print(f"  FAISS+OpenAI:        {faiss_results['query_time']:.2f}s")
    print(f"  ChromaDB+HuggingFace: {chroma_results['query_time']:.2f}s")

    if query_speedup > 1:
        print(f"  → FAISS is {query_speedup:.1f}x faster for queries")
    else:
        print(f"  → ChromaDB is {1/query_speedup:.1f}x faster for queries")

# Generate the detailed analysis
generate_detailed_recommendations(comparison_results)
````

````python
# ============================================================================
# SIDE-BY-SIDE RESPONSE COMPARISON (OPTIONAL)
# ============================================================================

def display_response_comparison(results):
    """Display detailed side-by-side response comparison"""

    print("\n" + "="*80)
    print("📝 DETAILED RESPONSE COMPARISON")
    print("="*80)

    if not all(result['success'] for result in results.values()):
        print("⚠️ Cannot compare responses - some implementations failed.")
        return

    faiss_answer = results['FAISS+OpenAI']['answer']
    chroma_answer = results['ChromaDB+HuggingFace']['answer']

    print(f"\n🔵 FAISS + OpenAI Response:")
    print("-" * 40)
    print(faiss_answer)
    print(f"\nResponse Length: {len(faiss_answer.split())} words")
    print(f"Sources Used: {results['FAISS+OpenAI']['source_docs_used']} documents")

    print(f"\n🟡 ChromaDB + HuggingFace Response:")
    print("-" * 40)
    print(chroma_answer)
    print(f"\nResponse Length: {len(chroma_answer.split())} words")
    print(f"Sources Used: {results['ChromaDB+HuggingFace']['source_docs_used']} documents")

    # Basic similarity analysis
    faiss_words = set(faiss_answer.lower().split())
    chroma_words = set(chroma_answer.lower().split())

    common_words = faiss_words.intersection(chroma_words)
    similarity_score = len(common_words) / len(faiss_words.union(chroma_words)) * 100

    print(f"\n📊 Response Similarity Analysis:")
    print(f"  Common words: {len(common_words)}")
    print(f"  Similarity score: {similarity_score:.1f}%")

    if similarity_score > 70:
        print("  → Very similar responses - both models retrieved relevant content")
    elif similarity_score > 50:
        print("  → Moderately similar - some differences in retrieval/generation")
    else:
        print("  → Different responses - may indicate different retrieval effectiveness")

# Display the side-by-side comparison
display_response_comparison(comparison_results)

print("\n" + "="*80)
print("🎉 RAG COMPARISON COMPLETE!")
print("="*80)
````
